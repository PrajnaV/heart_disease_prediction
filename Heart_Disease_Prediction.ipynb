Convert heart disease column values to boolean

```python
# Clean the 'Heart Disease' column to remove extra spaces and make lowercase
heart_data['Heart Disease'] = heart_data['Heart Disease'].str.strip().str.lower()
heart_data['Heart Disease'] = heart_data['Heart Disease'].map({'presence': 1, 'absence': 0})

```

Splitting features and target

```python
X = heart_data.drop(columns='Heart Disease',axis=1)
Y = heart_data['Heart Disease']
```
